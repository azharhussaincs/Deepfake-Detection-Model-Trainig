```python
# This script organizes raw video files into 'Real' and 'Fake' folders.
# It uses the global metadata JSON to determine the label of each video.
# It also implements balancing by limiting the number of fake videos 
# to match the number of real videos available in a given set.

import glob
import pandas as pd
import json
import os
import shutil 
```

```python
# Load the global metadata JSON file which maps filenames to FAKE/REAL labels
# Note: In the Kaggle dataset, each part has its own metadata.json. 
# This script assumes you are pointing to the metadata.json of the current part being processed.
path_to_json = '/home/albaloshi/Desktop/Deepfake_detection_using_deep_learning-master/Dataset/dfdc_train_part_0/metadata.json' # Example Metadata source path
with open(path_to_json) as f: # Open file for reading
  labels = json.load(f) # Load JSON content into dictionary
```

```python
# Convert the labels JSON into a Pandas DataFrame and transpose it
df = pd.DataFrame(labels).T # Transpose: keys (filenames) become rows
# Create separate lists for REAL and FAKE video filenames
read_df = pd.DataFrame(df.loc[df['label'] == 'REAL']) # Filter REAL videos
read_df_fake = pd.DataFrame(df.loc[df['label'] == 'FAKE']) # Filter FAKE videos

list_of_real = list(read_df.index.values) # Extract REAL filenames into list
list_of_fake = list(read_df_fake.index.values) # Extract FAKE filenames into list
print(f"Total REAL in metadata: {len(list_of_real)}, Total FAKE in metadata: {len(list_of_fake)}") # Log stats
```

```python
# Define the source directory containing raw videos (e.g., a specific Kaggle dataset part)
path = '/home/albaloshi/Desktop/Deepfake_detection_using_deep_learning-master/Dataset/dfdc_train_part_0/' # Example Input directory (change part number as needed)
video_files = glob.glob(path + "*.mp4") # List all MP4s in input directory
# Get only the base filenames
names = [] # Store extracted names
for i in video_files:
    # Use forward slash for Linux compatibility
    names.append(i.split("/")[-1]) # Get filename from path
```

```python
# Destination folders for organized data
src = path # Reference to source path
dest_f = '/home/albaloshi/Desktop/Deepfake_detection_using_deep_learning-master/Dataset/FF_FAKE/' # Consolidated Destination for Fakes
dest_r = '/home/albaloshi/Desktop/Deepfake_detection_using_deep_learning-master/Dataset/FF_REAL/' # Consolidated Destination for Reals

# Identify which videos in the source folder are REAL and which are FAKE
real_count = [] # List of REAL files found
fake_count = [] # List of FAKE files found
for i in names:
    if i in list_of_real: # If file is listed as REAL in metadata
        real_count.append(i) # Add to real list
    elif i in list_of_fake: # If file is listed as FAKE in metadata
        fake_count.append(i) # Add to fake list

print("Count of real in this folder:", len(real_count)) # Log local real count
print("Count of fake in this folder:", len(fake_count)) # Log local fake count
    
```

```python
# Copying REAL videos to their destination
for i in real_count:
    if "REAL" == (labels[i]['label']): # Double check label
        shutil.copyfile(src + i, dest_r + i) # Copy file to Real folder

# Copying FAKE videos, but only as many as there are REAL videos (Balancing)
for i in fake_count[:len(real_count)]: # Slice fake list to match real count
    if "FAKE" == (labels[i]['label']): # Double check label
        shutil.copyfile(src + i, dest_f + i) # Copy file to Fake folder

print("Organized files copied successfully.")
```
